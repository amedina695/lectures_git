```python
def sigrej(x, limits, mask=True):
    """
    Sigma clipping algorithm to remove bad pixels.

    Parameters
    ----------
    x : array_like
        Data.
    limits : float
        Threshold to reject pixels.
    mask : array_like of bool but optional
        Initial good pixel mask.
    
    Returns
    -------
    mask_out : ndarray of bool
        Modified boolean mask

    """
    x = np.asarray(x) # Converts input to array

    # Optiona boolean mask
    # If no mask, all data is good
    if mask is None:
        good = np.ones_like(x, dtype=bool)
    # If mask, should be same shape as array
    else:
        mask = np.asarray(mask, dtype=bool)
        if mask.shape != x.shape:
            raise ValueError("mask must have same shape as x")
        good = mask.copy()
    
    # Return modified boolean mask
    for N in limits:
        mean = np.mean(x[good])
        median = np.median(x[good])
        std = np.std(x[good], ddof=1)

        # Keep good points within N*sigma of the mean
        #low_limit  = median - N * std
        #high_limit = median + N * std
        #new_good = x[(low_limit < x) & (x < high_limit)]
        new_good = np.abs(x - mean) <= N * std
        good &= new_good

    return good
```
