```python
# Continuation of practicum 3
data_ccd       = np.zeros(400) # Empty array with 400 elements
n_p            = 10000
data_ccd[:396] = np.random.poisson(n_p, 396) # First 396 elements from Poisson dist

low            = 0
high           = 1e6
data_ccd[396:] = np.random.uniform(low, high, 4) # Last 4 elements from uniform dist
#print(data_ccd)
```

```python
# Sigma clipping
mean       = np.mean(data_ccd)
median     = np.median(data_ccd)
std        = np.std(data_ccd) # std of current data as limit
max_std    = 5 * std # Want a range of 5 sigma

# Range
range_low  = median - max_std
range_high = median + max_std

# Clipped array
clipped    = data_ccd[(range_low < data_ccd) & (data_ccd < range_high)]

#print(len(clipped))
#print(clipped)
```

```python
# New std and mean and median from practicum 3
# Now creating a subsample 
nmean       = np.mean(clipped)
nmedian     = np.median(clipped)
nstd        = np.std(clipped)
max_nstd    = 5 * nstd

# New range
nrange_low  = nmedian - max_nstd
nrange_high = nmedian + max_nstd

# New clipped array
nclipped    = clipped[(nrange_low < clipped) & (clipped < nrange_high)]

expected_std = np.sqrt(nmean)
```

```python
print("Old mean is:", mean,"; New mean is:", nmean)
print("Old median is:", median,"; New median is:", nmedian)
print("Old std is:", std,"; New std is:", nstd)
print("The expected mean of Poisson: ", expected_std,"\n")
```
