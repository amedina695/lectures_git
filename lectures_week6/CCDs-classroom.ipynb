### Charge-Coupled Devices (CCDs)

- silicon semiconductors used for imaging, photometry, spectroscopy, astrometry 
- best for VIS due to physical properties of silicon (e.g., WFC3 UV/VIS two 2k × 4k CCDs from 0.2-1.0$\mu$m; but uses IR Array for NIR)


- Detecting the final signal: **quantum efficiency (QE)**:
     $$\rm QE = \frac{\mathrm{N_{phot,detected}}}{\mathrm{N_{phot,incoming}}}$$
     - eye/sketchpad — ok for imaging, sensitivity bad and qualitative, cats QE 10%, human few %
     - photographic plates had QE 2% up to ~10%
     - photographic film — ok for imaging, but sensitivity poor; QE 10%
     - photomultipliers — 1D photometry only, accurate, fast, QE 50+%
     - array detectors — now almost all detectors except at radio and $\gamma$-ray wavelengths, QE 80-95% (e.g., https://jwst-docs.stsci.edu/jwst-near-infrared-spectrograph/nirspec-performance) CCDs generally QE> 80%; reaches 100% at some $\lambda$s (e.g., https://hst-docs.stsci.edu/wfc3ihb/chapter-5-wfc3-detector-characteristics-and-performance/5-4-wfc3-ccd-characteristics-and-performance)
     

#### How does it work?
- CCD is an array of metal insulator semiconductor (MIS) capacitors 
    - semiconductor chips sensitive to light
    - 2D array of "wells" (they produce pixels, and are sometimes even called pixels)
    
    <img src="ccdwell.png" width=300 height=300 />
    
- photons hit the CCD and may produce electrons (photoelectric effect)
    - due to silicon's structure this can happen for $\lambda_{\mathrm{phot}} \in [0.3, 1.1]\mu$m; outside this region CCD is "transparent" to light
    - recombination time ~100$\mu$s
- electrons get stored in depletion region of the capacitor
    - voltage allows electron to be moved in "gates"; potential wells where electrons can be stored till readout at end of exposure 
- in readout voltage changes on CCD so that electrons flow from one pixel to the other
    - not w/o losses of course; Charge Transfer Efficiency (CTE)
    - typical CTEs are 0.99999 to 0.999999, aka on average 1 photon lost per 100000 to 1000000 shifts
        - consider if CTE = 99.9% we would lose 1 photon per 1000 transfers...for >2000 transfers of a typical CCD most pixels would be non-readable
        
    <img src="ccdclocking.png" width=400 height=400 />
    - each column is connected in parallel $\rightarrow$ shifts happen simultaneously for all CCD 
    - top row gets moved off array into "output shift register": row of unexposed pixels
     <img src="ccdreadout.png" width=300 height=300 />
     
    - readout electronics count electrons in a serial way for all pixels of the shift register$\rightarrow$ translate to digital number $\rightarrow$ send to computer memory; repeat until shift register empty

    - Clock another row
    - Whole array reads in few sec, or fraction of sec
    - With tricks even huge CCD arrays can be read fast
        - e.g., HST UVIS 2$\times$ (2051×4096) arrays can be read out in 96s
        
    - Readout times depend on size and clocking rates chosen
        - Faster clocking $\rightarrow$ more noise in amplifier
        - Slower clocking $\rightarrow$ get closer to saturating
- **Gain**: the amount of photons/electrons registered needed to produce 1 digital count
    - typical values can be ~10$e^-$/count $\rightarrow$ for every 10$e^-$ we register 1 count
    


- On-chip amplifier used to help read signal has a bias voltage to attract the charge from the shift register
    - It also has a gain:
    
$$ V_{\mbox{out}} = gN_{e^{-}}+V_{\mbox{bias}} $$
    - $V_{\mbox{out}}$ digitized into ADU (Analog-to-Digital Units) or DN (Data Numbers)
    - Amplifiers are not perfect, pixel capacitance adds ``read noise''
    - Good CCDs have <5 $e^{-}$ read noise, or < 25 $e^{-}$ charge added per pixel


- **QE** 
    - depends strongly on wavelength and thickness of silicon 
<img src="hst_vis_uv.png" width=400 height=400 /> (from https://hst-docs.stsci.edu/wfc3ihb/chapter-5-wfc3-detector-characteristics-and-performance/5-4-wfc3-ccd-characteristics-and-performance)
        - assumed to be the same for all pixels of a CCD (almost true; necessary to do flat fielding!)
    - depends on T[K] 

- **Dark Current**: thermal vibrations in Si can free $e^{-}$ and get them trapped into pixel wells
    - during read-out these are part of the $e^{-}$ read as signal $\rightarrow$ noise 
    - strong T[K] dependence $$
\begin{equation}
I_{\mbox{d}} \sim e^{(-a/kT)}
\end{equation} $$ so CCDs usually cooled by liquid N$_{2}$ solution to ~77 K, small $\delta$T to avoid changes in dark current

### CMOS: Complementary Metal Oxide Semiconductor arrays
   - nearly identical to CCDs but they contain additional circuitry built into every pixel
   - can be cheaper
   - traditionally had smaller QE; newer ones rival CCD QEs (e.g., https://www.princetoninstruments.com/products/cosmos-family/cosmos/app-notes/advanced-cmos-detectors-for-astronomy)
   - faster read-outs due to their parallel read-out architecture

- 2 most common types in Astro: CCD and CMOS array, for visible and IR (CMOS becoming popular in digicams)
    - QE 80-95%, near theoretical max.
    - Quantitative, dimensionally near perfect
    - Formats vary, good optical CCD now 4k$\times$4k (e.g, Vera Rubin Observatory https://www.lsst.org/about/camera/features)

    - Big dynamic range, 12--24 meaningful bits common, 2-4 bytes/pixel
    
    - Monochromatic: must take 3 frames with different filters for true color
    - Pixel size 10--30 $\mu$m
    
    - There is a supplemental reading in *site-ccd.pdf* and *CCD\_Manual\_2010.pdf* for those interested in more about CCD chips and CCD observing.
    - Read PAN-STARRS web page on camera: http://www2.ifa.hawaii.edu/research/Pan-STARRS.shtml, https://outerspace.stsci.edu/display/PANSTARRS/PS1+The+Pan-STARRS1+facilities+and+hardware
    
    
- Recap 
    - All arrays:
         - Convert photons to electrons using a photosensitive material
         - Capture them in ``wells'' representing pixels in the image
         - Count the electrons by putting them through an amplifier and digitizer
    - CCD
         - Photosensitive material *is* the silicon that moves and counts the electrons
         - Electrons moved across the array
         - All pixels read through the same amplifier and digitizer
    - CMOS
         - Photosensitive material sits *on top* of the silicon that moves and counts the electrons
         - Electrons drop to a layer of capacitors
         - Usually multiple readout amplifiers and digitizers allowing for faster read out

```python
# Patch of 10 arcmin x 10 arcmin

x = 2048
y = 2048

pixels = 2048 * 2048 # pixels
arcmin = 600 * 600 # arcmin

resolution = arcmin / pixels # arcsec per pixel
print("The resolution is: {0:2f} arcsec/pixel".format(resolution))

# Moon aremin is 30 arcmin
moon_arc = 30 * 60 # arcsec
# For pixels its arcmin / R
moon_pixels = moon_arc / resolution
print("The number of pixels on the moon is: {0:2f}".format(moon_pixels))

#

prox_arc = 1.02 * 1e3
print("The number of pixels on the proxima is: {0:2f}".format(resolution / prox_arc))
```

- Suppose you collect 100 ADUs for an astronomical source. What is the uncertainty in that count (in ADUs) if the gain is 3 electrons per ADU?

### Think-pair-share (15 min)

- Assume that we observe a source that on its own registers 50 ADUs in a pixel in 60 sec of integration with our detector. Our detector has:
    - a gain of 2 e-/ADU
    - a dark current of 0.01 e-/pixel/sec
    - a read-out noise of 5 e-
    - assume a 0 bias.
    
    
- What is the SNR of our observation? How does it compare with the ideal SNR?

```python
gain = 2 # e- / ADU
ADU = 50 # ADU / pixel

total_e = gain * ADU # e- / pixel

dark_curr = 0.01 # e- / pixel /sec

readn = 5 # e-

t = 60 # seconds

total_dark = t * dark_curr # e- / pixel

signal = total_e - total_dark

print(signal)

noise = np.sqrt( total_e * dark_curr * t * readn)
print("SNR of: ", signal/noise)
```

### Assuming you are using a CCD with a single read out configuration and a charge transfer efficiency of 99.0%. Let's study how much signal losses you have from the last row to be read in, in  a 2048 × 2048 array. Determine how much charge (%) you will lose from the first pixel, middle (1024) and last (2048) pixel of the last row to be readout. 

### What about when you have a CTE of 99.9% , 99.99%, or 99.999% efficiency? Determine what CTE is needed in order to produce an acceptable loss of charge for this same array. (based on exercise from Howell )

```python
import numpy as np

# Setup

x = 2048
y = 2048

cte = 0.99
Nphot = 1e7

N1 = Nphot * cte

print('First pixel (one transfer) will record {0:2f} photons and lost{1:2f}'.format(N1, N1 - Nphot))
```

```python
# THe first pixel from the last row will have read out

N2 = Nphot
for i in range(y):
    N2 *= cte

print('First pixel of last row will record {0:2f} photons and lost{1:2f}'.format(N2, N2 - Nphot))
```

```python
# For the middle row

N3 = Nphot
for i in range(y+int(x/2)):
    N3 *= cte
```

```python
def compute_efficiency(y, N_photon, N_effic):

    N = N_photon
    for i in range(y):
        N *= N_effic
    
    return N
```

```python
Nse = [.999, .9999, .99999, 1]

for i in Nse:
    out = compute_efficiency(y, Nphot, i)
    print("The efficiency {0:2f} for last pixel has output of {1:2f} and loss of {2:2f}".format(i, out, 100 - np.abs(out-Nphot)))
```
