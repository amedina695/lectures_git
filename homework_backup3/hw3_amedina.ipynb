```python
def squareplot(low, high, n, saveplot=False):

    x = np.linspace(low, high, n)
    y = sup.square(x)

    plt.figure()
    plt.plot(x, y)
    plt.title("Square Function")
    plt.xlabel("Input")
    plt.ylabel("Output")

    if saveplot:
        plt.savefig(f"{saveplot}.pdf", format="pdf")
        plt.show()
    else:
        plt.show()
```

```python
squareplot(1, 10, 11)
```
