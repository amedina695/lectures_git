```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Creating a square function

def square(x):
    """A function that takes an input of a scalar
    or an array if any dimension or type and returns
    the square.

    Parameters
    -----------
    x: int or array_like

    Returns
    -------
    int or array
    """

    # Looking at numpy documentation
    # Can't just return square but
    # https://numpy.org/doc/2.2/reference/generated/numpy.asarray.html
    x = np.asarray(x) # Converts input to array
    return x**2
```

```python
print(square(9))
y = np.arange(1, 10)
print(y)
print(square(y))
```
