```python
# Using my version of the code

def create_samp(N):
    n_times = 10
    mu = 55
    sigma = 13

    res = []

    for i in range(n_times):
        y = np.random.normal(mu, sigma, N)
        s_std = np.std(y)
        s_mean = np.mean(y)
        res.append([i+1, s_mean, s_std])
    
    x = np.array(results)
    
    return x
```

```python
read_samp_data = np.loadtxt('practicum2_am1.txt')

sigma_of_means = np.zeros((6, 2))

array_size = [10, 100, 1000, 10000, 100000, 1000000]

for j in range(6):
    sigma_of_means[ j, 0 ] = array_size[j]

    # get the means and their std
    sigma_of_means[ j, 1 ] = np.std(read_samp_data[j * 10 : (j+1) * 10, 1])


for j in range(6):
    print('{0:f}{1:.5f}'.format(sigma_of_means[j, 0], sigma_of_means[j, 1]))

with open('practicum2_am1.txt', 'ab') as f:
    f.write(b'\n')
    np.savetxt(f, sigma_of_means, fmt='%d %.4f', header='Sample size Sigma of means')
```

Problem 6: AI Version

```python
file_path = "practicum2_am1.txt"

# Storage for results
sample_size_std = {}

with open(file_path, 'r') as f:
    lines = f.readlines()

current_sample_size = None
current_means = []

for line in lines:
    line = line.strip()
    if line.startswith("# Number of draws in this sample: "):
        # If we were collecting means, compute std and store
        if current_sample_size and current_means:
            std_of_means = np.std(current_means)
            sample_size_std[current_sample_size] = std_of_means
            current_means = []

        # Extract new sample size
        current_sample_size = int(line.split(":")[1].strip())

    elif line and not line.startswith("#"):
        # Parse the line assuming format: index mean std
        parts = line.split()
        if len(parts) == 3:
            mean = float(parts[1])
            current_means.append(mean)

# Handle last block
if current_sample_size and current_means:
    std_of_means = np.std(current_means)
    sample_size_std[current_sample_size] = std_of_means

# Print results
for size, std in sample_size_std.items():
    print(f"Sample size {size}: Std of means = {std:.2f}")
```
