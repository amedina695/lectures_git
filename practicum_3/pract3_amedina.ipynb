```python
# Using and importing the linfit

import linfit as L

# it goes linfit(y, x, y_unc) where y_unc is y uncertainty
```

```python
a, b, sa, sb, chisq, prob, covar, yfit = L.linfit(y1, x1, 0.2) #np.repeat(0.9, len(y1)))
#print(yfit)


print('# a = {0:.3f} +- {1:.3f}'.format(a, sa))  # Only believe these if the rest makes sense!
print('# b = {0:.3f} +- {1:.3f}'.format(b, sb))  #   "
print('# chisq = {0:.3f}'.format(chisq))        # Chi-squared should be close to N-m (degrees of freedom)
print('# prob  = {0:.3f}'.format(prob) )        # Even 0.1 is believable.  Much lower (e.g., 0.001) is not!
print('# covar = ')                             # Covariance matrix: uncertainties = sqrt(diag(covar))
print('# ' + str(covar))
print('# reduced chisq = {0:.3f}'.format(chisq / (x1.size - 2)))  # Chi-squared / (N-m) should be around 1
```

We are within 3 sigma of our fit. Reduced chi squared decreases with increasing uncertainty, and the uncertainties in a and b also increase.

The linfit does not for the data because it assumes the data has a linear fit. To fix this, we can use an alternative fit method, like increasing the degree of the fitted function

```python
import scipy.optimize as spo
from scipy.optimize import curve_fit

def trycurvefit(x, a, b):
    return a*x**2 + b*x

(fit, cov_x, infodict, mesg, ier) = curve_fit(trycurvefit, x2, y2, full_output=True, sigma=0.5)
fiterr = np.sqrt(np.diag(cov_x))
```

```python
# check status FIRST!
print("Status: {0:d}".format(ier)) # 1 = solution found
print(mesg)
print("Reduced chi-squared:{0:.4e}".format(np.sum(trycurvefit(x1, *fit) - y1)**2 / (len(x1) - fit.size))) # reduced chi-squared
print("Results and uncertainties:")
print(np.array((fit, fiterr)).T) # ok, now the results
print("Covariance matrix:")
print(cov_x)  # and errors, including covariance terms
print("# function evals: {0:d}".format(infodict['nfev']))
```

```python
# Array of 400 elements
array = np.zeros(400)

# the first 396 are chosen from a Poisson distribution

n_p        = 10000
samples_p = 396

p = np.random.poisson(n_p, samples_p)
#print(p)
array = np.append(array, p)

# Drawing from a uniform distribution for only 4 points

n_u_low = 0
n_u_upp = 1000000

samples_u = 4

u = np.round(np.random.uniform(n_u_low, n_u_upp, samples_u))
#print(u)

array = np.append(array, u)

print(len(array))
```

```python
data_ccd = np.zeros(400)
data_ccd[:396] = np.random.poisson(n_p, 396)

data_ccd[396:] = np.random.uniform(n_u_low, n_u_upp, samples_u)

mean = np.mean(data_ccd)
print(mean)

median = np.median(data_ccd)
print(median)
```

```python
# Finding sigma and creating a mask to include only point witihn 5 sigma of the median

std = np.std(data_ccd)
print(std)

max_std = 5 * std

range_low = median - max_std
range_high = median + max_std


clipped = data_ccd[(range_low < data_ccd) & (data_ccd < range_high)]

print(len(clipped))
print(clipped)
```

```python
# New std and mean and median

nmean = np.mean(clipped)
nmedian = np.median(clipped)
nstd = np.std(clipped)

print(nmean, nmedian, nstd)
```
