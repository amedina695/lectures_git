```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import astropy.io.fits as fits
from scipy.optimize import curve_fit
```

```python
# Problem 1a
x, dx = 5.04, 0.02
y, dy = 15.3, 0.3

z = x / y

par1 = 1 / y
par2 = - x / y**2

dz = np.sqrt( (par1 * dx)**2 + (par2 * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.4f}'.format(z, dz))
```

```python
# Problem 1b
x, dx = 2.12, 0.48
y, dy = 0.31, 0.02

z = x * y**3

par1 = y**3
par2 = 3 * x * y**2

dz = np.sqrt( (par1 * dx)**2 + (par2 * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z, dz))
```

```python
# Problem 1c
x, dx = 0.22, 0.14
y, dy = 1.23, 0.02

z = x * np.log10(2 * y)

par1 = np.log10(2 * y)
par2 = x / y

dz = np.sqrt( (par1 * dx)**2 + (par2 * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z, dz))
```
