```python
x, dx = 3.1, 0.2
y, dy = 2.31, 0.32

z = 5 * x**(1/2) + 2*y**3

dzdx = (5 / 2) * x**(-1/2)
dzdy = 6 * y**2

dz = np.sqrt( (dx * dzdx)**2 + (dy * dzdy)**2 )

print(f'z +_ dz = {z} + {dz}')
```
