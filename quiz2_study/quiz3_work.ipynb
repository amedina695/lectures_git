```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.io.fits as fits
from scipy.optimize import curve_fit
```

```python
# 1

x, dx = 0.53, 0.12
y, dy = 4.2, 0.4

# Equation
z1 = x**2

# Partial derivative w.r.t first variable
dz1dx = 2 * x

# To find the error probagation

dz1 = np.sqrt( (dz1dx * dx)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z1, dz1))
```

```python
# Equation
z2 = 2 * x * y**2

# Partial derivative w.r.t first variable
dz2dx = 2 * y**2
dz2dy = 4 * x * y

# To find the error probagation

dz2 = np.sqrt( (dz2dx * dx)**2 + (dz2dy * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z2, dz2))
```

```python
# Equation
z3 = x * np.exp(2 * y) + 2 * y**2

# Partial derivative w.r.t first variable
dz3dx = np.exp(2 * y)
dz3dy = 2 * x * np.exp(2 * y) + 4 * y

# To find the error probagation

dz3 = np.sqrt( (dz3dx * dx)**2 + (dz3dy * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z3, dz3))
```
