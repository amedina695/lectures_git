```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = ([1, 5, 2, 1, 3, 4, 9])

dmean = np.mean(data)
print("The mean is the average which is:", dmean)

dmedian = np.median(data)
print("The median is the midpoint whici is given by:", dmedian)

dvar = np.var(data, ddof=1)
print("The variance is the std squared:", dvar)

dstd = np.std(data, ddof=1)
print("The standard deviaation is:", dstd)
```

```python
# Problem 3.1

x, dx = 5.1, 0.4
y, dy = 3.2, 0.1

z = x * y**2

dzdx = y**2
dzdy = 2*x*y

dz = np.sqrt( (dzdx * dx)**2 + (dzdy * dy)**2 )

print('z +- dz = {0:.2f}+-{1:.2f}'.format(z, dz))
```

```python
# Problem 3.2

a, da = 3.1, 0.3
b, db = 2.1, 0.1
c, dc = 4.36, 0.12

# Equation
d = a * ( 1 / b ) - c

# Partials
ddda = ( 1 / b )
dddb = a * ( -1 / b**2 )
dddc = -1

# d error
dd = np.sqrt( (ddda * da)**2 + (dddb * db)**2 + (dddc * dc)**2)

print('d +- dd = {0:.2f}+-{1:.2f}'.format(d, dd))
```

```python
# Problem 3.3

l, dl = 1.2, 1.2
p, dp = 0.32, 0.01

# Equation
r = l * np.sqrt(p)

# Partials
drdl = np.sqrt(p)
drdp = l * (1/2) * p**(-1/2)

# r error
dr = np.sqrt( (drdl * dl)**2 + (drdp * dp)**2 )

print('r +- dr = {0:.2f}+-{1:.2f}'.format(r, dr))
```
